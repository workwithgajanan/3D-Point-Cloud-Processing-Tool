```python
def batch_process_las_files(folder_path):
    # Folder madhlya saglya .las files shodhne
    search_pattern = os.path.join(folder_path, "*.las")
    las_files = glob.glob(search_pattern)
    
    if not las_files:
        print("Folder madhye ek hi .las file sapadli nahi.")
        return

    report_lines = []
    report_lines.append(f"--- 3D Point Cloud Batch Processing Report ---")
    report_lines.append(f"Generated on: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
    report_lines.append(f"Total files found: {len(las_files)}\n")
    report_lines.append("-" * 50)

    print(f"Total {len(las_files)} files sapadlya. Processing chalu ahe...\n")

    # Pratyek file loop madhye process karne
    for file_path in las_files:
        file_name = os.path.basename(file_path)
        try:
            las = laspy.read(file_path)
            point_count = len(las.points)
            z_min, z_max = np.min(las.z), np.max(las.z)
            
            report_lines.append(f"File: {file_name}")
            report_lines.append(f"Status: SUCCESS")
            report_lines.append(f"Total Points: {point_count}")
            report_lines.append(f"Height Range (Z): {z_min:.2f} to {z_max:.2f}")
            report_lines.append("-" * 50)
            
            print(f"Processed: {file_name}")

        except Exception as e:
            report_lines.append(f"File: {file_name}")
            report_lines.append(f"Status: FAILED")
            report_lines.append(f"Error Message: {e}")
            report_lines.append("-" * 50)
            print(f"Failed: {file_name}")

    # Sagla data Report.txt madhye save karne
    report_filename = "Summary_Report.txt"
    with open(report_filename, "w") as file:
        file.write("\n".join(report_lines))
        
    print(f"\nAutomation Complete! Check '{report_filename}' in your folder.")

# Main Execution
if __name__ == "__main__":
    # '.' mhanje current folder jithe tumchi ipynb file ahe
    current_directory = "." 
    batch_process_las_files(current_directory)
```
