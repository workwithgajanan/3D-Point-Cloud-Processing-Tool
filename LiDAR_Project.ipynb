```python
import laspy
import numpy as np
import glob
import os
from datetime import datetime
```

```python

def analyze_las_file(file_path):
    print(f"--- Analyzing File: {file_path} ---")
    try:
        # File read karne
        las = laspy.read(file_path)
        
        # Total points count karne
        point_count = len(las.points)
        print(f"Total Points: {point_count}")
        
        # X, Y, Z coordinates baher kadhne
        x, y, z = las.x, las.y, las.z
        
        # Data chi range print karne
        print(f"X coordinate range : {np.min(x)} to {np.max(x)}")
        print(f"Y coordinate range : {np.min(y)} to {np.max(y)}")
        print(f"Z (Height) range   : {np.min(z)} to {np.max(z)}")
        
        print("Success: File successfully processed!\n")
        
    except FileNotFoundError:
        print(f"Error: '{file_path}' navachi file sapadli nahi. File name check kara.\n")
    except Exception as e:
        print(f"Error aala: {e}\n")

# Main block
if __name__ == "__main__":
    # Tumchya file cha nav ithe taka
    file_name = "simple.las" 
    analyze_las_file(file_name)
```
